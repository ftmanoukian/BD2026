# Ejemplo en clase: usar pandas para responder preguntas simples

En este notebook vamos a usar un dataset real de la Ciudad de Buenos Aires.

La idea no es hacer un análisis estadístico avanzado, sino practicar este recorrido:

**archivo de datos → tabla → pregunta sencilla → conteo → gráfico → conclusión**

## 1. Importamos pandas

`pandas` es una biblioteca de Python que permite trabajar con tablas de datos.

````python
import pandas as pd
````

## 2. Cargamos el archivo CSV

Primero subimos el archivo `arbolado-publico-lineal.csv` a Colab usando el botón de archivos.

Después lo leemos con `pd.read_csv()`.

````python
df = pd.read_csv("arbolado-publico-lineal.csv")
````

## 3. Miramos las primeras filas

`head()` muestra las primeras filas de la tabla.

Pregunta importante:

**¿Qué representa cada fila?**

````python
df.head()
````

En este dataset, cada fila representa un árbol registrado en la vía pública.

## 4. Miramos el tamaño de la tabla

`shape` nos dice cuántas filas y columnas tiene el dataset.

````python
df.shape
````

## 5. Miramos los nombres de las columnas

Esto sirve para elegir qué datos podemos usar.

````python
df.columns
````

Para este ejemplo vamos a usar columnas simples como:

- `comuna`
- `nombre_com`
- `calle_nombre`

## 6. Pregunta 1

**¿Hay la misma cantidad de árboles registrados en todas las comunas?**

Para responder, contamos cuántas veces aparece cada comuna.

````python
arboles_por_comuna = df["comuna"].value_counts()

arboles_por_comuna
````

Ahora hacemos un gráfico de barras.

````python
arboles_por_comuna.plot(kind="bar", title="Cantidad de árboles registrados por comuna")
````

### Conclusión 1

Si algunas barras son mucho más altas que otras, entonces los árboles registrados **no están distribuidos de manera pareja entre comunas**.

La conclusión debe escribirse con cuidado:

> En este dataset, algunas comunas tienen más árboles registrados que otras.

## 7. Pregunta 2

**¿Cuáles son las especies de árboles más frecuentes?**

Para no hacer un gráfico enorme, mostramos solo las 10 especies más comunes.

````python
especies_mas_comunes = df["nombre_com"].value_counts().head(10)

especies_mas_comunes
````

````python
especies_mas_comunes.plot(kind="barh", title="10 especies más frecuentes")
````

### Conclusión 2

Si unas pocas especies aparecen muchas veces, podemos decir:

> En este dataset, algunas especies aparecen con mucha más frecuencia que otras.

## 8. Pregunta 3

**¿En qué calles aparecen más árboles registrados?**

````python
calles_con_mas_arboles = df["calle_nombre"].value_counts().head(10)

calles_con_mas_arboles
````

````python
calles_con_mas_arboles.plot(kind="barh", title="10 calles con más árboles registrados")
````

## 9. Una hipótesis que quizá no se cumple

A veces nuestra idea inicial no coincide con los datos.

Por ejemplo:

> Hipótesis: la Comuna 1 es la que tiene más árboles registrados.

Para verificarlo, volvemos al conteo por comuna y miramos cuál aparece primera.

````python
arboles_por_comuna.head(5)
````

### Conclusión posible

Si la Comuna 1 no aparece primera, la hipótesis **no se cumple**.

Eso no está mal: una hipótesis puede cumplirse o no cumplirse. Lo importante es poder justificar la respuesta con datos.
